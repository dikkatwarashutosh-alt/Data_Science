```python
import numpy as np
```

```python
# creatning a simple array

arr1=np.array([10,20,30,40,50])
print(arr1)
```

```python
# creating a 2d array

arr2=np.array([[1,2,3,4],[4,5,6,7],[7,8,9,1]])
print(arr2)
```

```python
# creating a 3d array

arr3=np.array([[[1,2,3],[4,5,6]],[[3,2,1],[4,5,6]],[[6,4,2],[7,2,6]]])
print(arr3)
```

```python
# dtype

arr1=np.array([2,1,3,4],dtype=float)
print(arr1)
print(arr1.dtype)
```

```python
# arange()

print(np.arange(1,10))
print(np.arange(1,10,2))
```

```python
# reshape()

print(arr2)
print("After reshaping the array")
print(arr2.reshape(2,6))  #2 row and 6 column
print("\n\n")

# reshape of 3d array

print(arr3)
print("After reshaping the 3d array")
print(arr3.reshape(2,3,3))  # 2-> two 2d array, 3,3-> 3 row & 3 column
```

```python
# np.ones and np.zeros

arr_zero=np.zeros((3,5))  # creating 3 * 5 matrix with 0 values only 
print(arr_zero,"\n")

arr_ones= np.ones((3,4))  # creating 3 * 4 matrix with 1 values only
print(arr_ones)
```

```python
#np.random

arr_random_values=np.random.random((3,3))   # creating 3*3 matrix with random values
print(arr_random_values)
```

```python
# np.linspace

print(np.linspace(1,15,6))
print(np.linspace(1,15,6,dtype=int))   # creating 6 values between 1 to 5 with equal distance
```

```python
#np.identity

print(np.identity(4))  #printing 4*4 identity matrix
```

```python
#np.full

np.full((2,3),18)  # creating 2*3 matrix with 18 values only
np.full((2,2,2),18)  # creating 2*2*2 tensor/3d array with 18 values only
```

```python
# Array attributes

# ndim

print(arr1)
print(arr1.ndim) # printing the dimention of the array
print("---------------------")
print(arr2)
print(arr2.ndim)
print("---------------------")
print(arr3)
print(arr3.ndim)
```

```python
# shape 

print(arr1)
print(arr1.shape) # printing the shape of the array
print("---------------------")
print(arr2)
print(arr2.shape)
print("---------------------")
print(arr3)
print(arr3.shape)
```

```python
# size

print(arr1.size)  #printing total number of element in array
print(arr2.size)
print(arr3.size)
```

```python
# itemsize

print(arr1.itemsize)  # printing the size of each element in array
print(arr2.itemsize)
```

```python
# dtype 

print(arr1.dtype)  # printing the data type of each element in array
```

```python
# changing the data type of array

print(arr1)
print(arr1.dtype)
print("-------------")
print(arr1.astype(float))
print(arr1.astype(np.int32))
```

```python
# Array Operations

a=np.arange(10)
b=np.arange(10)
```

```python
# scalar operations 

print(a)
print(a+2)
print(a-2)
print(a*2)
```

```python
# relational oprations

print(b<5)
print(a==b)
```

```python
# vector operations

print(a+b)
a-b
a*b
a**b
```
