```python
import numpy as np
```

# Array Creation

```python
# creatning a simple array

arr1=np.array([10,20,30,40,50])
print(arr1)
```

```python
# creating a 2d array

arr2=np.array([[1,2,3,4],[4,5,6,7],[7,8,9,1]])
print(arr2)
```

```python
# creating a 3d array

arr3=np.array([[[1,2,3],[4,5,6]],[[3,2,1],[4,5,6]],[[6,4,2],[7,2,6]]])
print(arr3)
```

# Basic Methods

```python
# arange()

print(np.arange(1,10))
print(np.arange(1,10,2))
```

```python
# reshape()

print(arr2)
print("After reshaping the array")
print(arr2.reshape(2,6))  #2 row and 6 column
print("\n\n")

# reshape of 3d array

print(arr3)
print("After reshaping the 3d array")
print(arr3.reshape(2,3,3))  # 2-> two 2d array, 3,3-> 3 row & 3 column
```

```python
# np.ones and np.zeros

arr_zero=np.zeros((3,5))  # creating 3 * 5 matrix with 0 values only 
print(arr_zero,"\n")

arr_ones= np.ones((3,4))  # creating 3 * 4 matrix with 1 values only
print(arr_ones)
```

```python
#np.random

arr_random_values=np.random.random((3,3))   # creating 3*3 matrix with random values
print(arr_random_values)
```

```python
# np.linspace

print(np.linspace(1,15,6))
print(np.linspace(1,15,6,dtype=int))   # creating 6 values between 1 to 5 with equal distance
```

```python
#np.identity

print(np.identity(4))  #printing 4*4 identity matrix
```

```python
#np.full

np.full((2,3),18)  # creating 2*3 matrix with 18 values only
np.full((2,2,2),18)  # creating 2*2*2 tensor/3d array with 18 values only
```

# Array Functions

```python
a=np.arange(9).reshape(3,3)
a
```

```python
# max/min/sum/prod
# 0 -> col and 1 -> row

print(np.max(a,axis=0))
print(np.min(a,axis=1))
print(np.sum(a))
print(np.prod(a,axis=1))
```

```python
# mean/median/std/var

print(np.mean(a))
print(np.median(a,axis=0))
```

```python
# trigonomoetric functions

np.sin(a)
```

```python
# Dot Production

x=np.arange(6).reshape(2,3)
y=np.arange(6).reshape(3,2)
print(x)
print(y)

print(np.dot(x,y))

# note 1st matrix column == 2nd matrix row count in this (2,3)&(3,2)
```

```python
# round/floor/ceil

print(np.round([3.5,3.4]))  # if .5 or more then next number as normal we do
print(np.floor([3.5,3.4]))  # it will take only intiger value like 3.4 = 3
print(np.ceil([3.5,3.4]))   # if 3.4, 3.2 ,3.5 ,3.6 then it take 4 for all
```

# Reshaping

```python
# reshape()

reshaping_demo=np.arange(12).reshape(4,3)
print(reshaping_demo)
```

```python
# transpose

print(np.transpose(reshaping_demo)) # it convert row into column and column into row
```

```python
# ravel

print(np.ravel(reshaping_demo))  # it convert 2d and 3d array into 1d array
print(reshaping_demo.ravel())   # another way to use ravel()
```

# Stacking

```python
# horizontal stacking

a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
```

```python
print(np.hstack((a4,a5)),"\n")
print(np.hstack((a4,a5,a4,np.arange(9).reshape(3,3))))
```

```python
# vertical stacking

print(np.vstack((a5,a4)))
```

# Splitting

```python
# horizotal split

print(a5)
print(np.hsplit(a5,2)) # it will split horizontaly in 2 parts
```

```python
# vertical split

print(np.vsplit(a5,3))  # array split verticaly in 3 part
```
